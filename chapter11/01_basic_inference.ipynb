# 第11章 基本的な推論

サンプル画像を準備し、CLIとPythonの両方で物体検出を実行します。

## 実行準備

次のセルで、推論に使用するサンプル画像を準備します。

```python
from pathlib import Path
from shutil import copyfile
from urllib.request import urlretrieve

if not Path("image.jpg").exists():
    urlretrieve("https://ultralytics.com/images/bus.jpg", "image.jpg")

Path("images").mkdir(exist_ok=True)
for index in range(1, 4):
    copyfile("image.jpg", f"image{index}.jpg")
    copyfile("image.jpg", f"images/image{index}.jpg")
```

## CLIによる推論

```python
!yolo detect predict model=yolo26n.pt source=https://ultralytics.com/images/bus.jpg
```

## 基本的な推論

```python
from ultralytics import YOLO

# モデルをロード
model = YOLO('yolo26n.pt')

# 画像に対して推論を実行
results = model('image.jpg')

# 検出数を表示
print(f"{len(results[0].boxes)}個の物体を検出しました")
```

## 結果の保存

```python
# 推論と同時に結果を保存
results = model('image.jpg', save=True)
```

## 保存先を指定

```python
results = model('image.jpg')
results[0].save('my_result.jpg')
```

## 結果の表示

```python
results = model('image.jpg')
results[0].show()  # ウィンドウで結果を表示
```

## パラメータの指定

```python
# 信頼度閾値を0.5に設定
results = model('image.jpg', conf=0.5)

# 複数のパラメータを同時に指定
results = model(
    'image.jpg',
    conf=0.5,      # 信頼度閾値
    iou=0.3,       # NMS閾値
    imgsz=640      # 画像サイズ
)
```

## 複数画像への推論

```python
# 複数の画像をまとめて推論
image_paths = ['image1.jpg', 'image2.jpg', 'image3.jpg']
results = model(image_paths)

# 各結果を処理
for i, result in enumerate(results):
    print(f"{image_paths[i]}: {len(result.boxes)}個検出")
```
