# 第12章 学習の再開と実験管理

中断した学習の再開、出力先の整理、Weights & Biasesによる実験管理を扱います。

## 学習の再開

```python
from ultralytics import YOLO

# 中断された学習を再開
model = YOLO("runs/detect/train/weights/last.pt")
model.train(resume=True)
```

## パラメーターを上書きして再開

```python
model = YOLO("runs/detect/train/weights/last.pt")
model.train(
    resume=True,
    batch=8,        # バッチサイズを変更（メモリ不足対策）
    patience=100,   # 早期終了の設定を変更
    device=0        # 使用するGPUを変更
)
```

## 実験名の設定

```python
from ultralytics import YOLO

# ベースライン実験
model = YOLO("yolo26n.pt")
model.train(
    data="coco128.yaml",
    epochs=100,
    name="baseline"
)
# → 結果は runs/detect/baseline に保存

# 学習率を変更した実験
model = YOLO("yolo26n.pt")
model.train(
    data="coco128.yaml",
    epochs=100,
    lr0=0.001,
    name="lr_0.001"
)
# → 結果は runs/detect/lr_0.001 に保存
```

## プロジェクトごとの整理

```python
# プロジェクトA：カスタムデータセットの学習
model = YOLO("yolo26n.pt")
model.train(
    data="custom_dataset.yaml",
    epochs=100,
    project="project_A",
    name="baseline"
)
# → 結果は runs/detect/project_A/baseline に保存
```
