# 第12章 学習オプション

エポック数やバッチサイズなど、学習条件ごとの設定例です。必要な設定を選んで実行してください。

## 実行準備

次のセルで、各設定例に使用するモデルを読み込みます。

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")
```

## エポック数の指定

```python
# 100エポック学習
model.train(data="coco128.yaml", epochs=100)

# 300エポック学習
model.train(data="coco128.yaml", epochs=300)
```

## バッチサイズの指定

```python
# 自動設定
model.train(data="coco128.yaml", epochs=100, batch=-1)

# 手動で指定
model.train(data="coco128.yaml", epochs=100, batch=16)
```

## 画像サイズの指定

```python
# 640×640（デフォルト）
model.train(data="coco128.yaml", epochs=100, imgsz=640)

# より高解像度（メモリに注意）
model.train(data="coco128.yaml", epochs=100, imgsz=1280)
```

## デバイスの指定

```python
# 自動選択（デフォルト）：利用可能なGPUを自動検出
model.train(data="coco128.yaml", epochs=100)

# GPU 0を使用
model.train(data="coco128.yaml", epochs=100, device=0)

# 複数GPU（0と1）を使用
model.train(data="coco128.yaml", epochs=100, device=[0, 1])

# CPUを使用（非推奨：非常に遅い）
model.train(data="coco128.yaml", epochs=100, device="cpu")
```

## 保存先の指定

```python
# デフォルト：runs/detect/train, train-2, train-3...
model.train(data="coco128.yaml", epochs=100)

# カスタムプロジェクトとラン名を指定
model.train(
    data="coco128.yaml",
    epochs=100,
    project="my_experiments",
    name="baseline_yolo26n"
)
# → 結果は runs/detect/my_experiments/baseline_yolo26n に保存
```

## 上書き設定

```python
# 既存ディレクトリを上書き
model.train(
    data="coco128.yaml",
    epochs=100,
    name="my_experiment",
    exist_ok=True
)
# 同じnameで再実行しても、my_experimentが上書きされる
```

## 最適化アルゴリズムの指定

```python
# 自動選択（デフォルト、推奨）
model.train(data="coco128.yaml", epochs=100, optimizer="auto")

# 手動で指定
model.train(data="coco128.yaml", epochs=100, optimizer="AdamW")
model.train(data="coco128.yaml", epochs=100, optimizer="SGD")
```

## 学習率の指定

```python
model.train(
    data="coco128.yaml",
    epochs=100,
    optimizer="AdamW",  # 学習率を手動設定する場合は明示的に指定
    lr0=0.01,           # 初期学習率
    lrf=0.01,           # 最終学習率（lr0の1%まで減衰）
    cos_lr=False        # 線形減衰（デフォルト）
)
```

## 早期終了の設定

```python
# デフォルト：patience=100
model.train(data="coco128.yaml", epochs=300)

# より早く終了する場合
model.train(data="coco128.yaml", epochs=300, patience=50)
```

## 基本的なデータ拡張の設定

```python
model.train(
    data="coco128.yaml",
    epochs=100,
    # 色調変換
    hsv_h=0.015,      # 色相の変化範囲（0.0〜1.0）
    hsv_s=0.7,        # 彩度の変化範囲（0.0〜1.0）
    hsv_v=0.4,        # 明度の変化範囲（0.0〜1.0）
    # 幾何学変換
    degrees=0.0,      # 回転角度（-180〜180）
    translate=0.1,    # 平行移動の範囲（0.0〜1.0）
    scale=0.5,        # スケール変化の範囲（0.0〜1.0）
    shear=0.0,        # せん断変換の角度（度）
    perspective=0.0,  # 透視変換の強度（0.0〜0.001程度）
    # 反転
    fliplr=0.5,       # 左右反転の確率（0.0〜1.0）
    flipud=0.0,       # 上下反転の確率（0.0〜1.0）
    # その他
    bgr=0.0           # RGB↔BGRチャンネル入れ替え確率（0.0〜1.0）
)
```

## 高度なデータ拡張の設定

```python
model.train(
    data="coco128.yaml",
    epochs=100,
    mosaic=1.0,       # Mosaicを使用する確率（0.0〜1.0）
    mixup=0.0,        # MixUpを使用する確率（0.0〜1.0）
    cutmix=0.0        # CutMixを使用する確率（0.0〜1.0）
)
```
