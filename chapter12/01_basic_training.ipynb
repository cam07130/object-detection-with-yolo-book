# 第12章 基本的な学習

学習済みモデルを利用した学習、ゼロからの学習、COCO128を使った学習を実行します。

## COCO学習済みモデルの自動ダウンロード

````python
from ultralytics import YOLO

# YOLO26nモデルをロード（初回はCOCO学習済み重みを自動ダウンロード）
model = YOLO("yolo26n.pt")

# カスタムデータセットでファインチューニング
model.train(data="custom_data.yaml", epochs=100)
````

## ゼロからの学習（非推奨）

````python
# ゼロから学習
model = YOLO("yolo26n.yaml")
model.train(data="custom_data.yaml", epochs=300)
````

## COCO128の自動ダウンロード

````python
from ultralytics import YOLO

# モデルのロード
model = YOLO("yolo26n.pt")

# COCO128での学習(初回実行時に自動ダウンロード)
results = model.train(data="coco128.yaml", epochs=3)
````

## data.yamlの基本構造

この例は概念的な設定ファイルなので、実行セルではなくYAMLとして掲載します。

```yaml
# データセットのパス
path: coco128  # データセットのルートディレクトリ
train: images/train2017    # 学習用画像のパス（pathからの相対パス）
val: images/train2017      # 検証用画像のパス（COCO128では学習と同じ）

# クラス名のリスト
names:
  0: person
  1: bicycle
  2: car
  3: motorcycle
  ...
  79: toothbrush
```


## 最初の学習実行

````python
from ultralytics import YOLO

# YOLO26nモデルをロード(COCO学習済み重み)
model = YOLO("yolo26n.pt")

# COCO128で3エポック学習
results = model.train(
    data="coco128.yaml",  # データセット設定ファイル
    epochs=3,              # 学習エポック数
    imgsz=640              # 入力画像サイズ
)
````
