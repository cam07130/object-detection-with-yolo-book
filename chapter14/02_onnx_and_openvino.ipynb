# 第14章 ONNXとOpenVINO

モデルをONNXやOpenVINOへ変換し、変換後のモデルで推論と速度比較を行います。

## 基本的なONNXエクスポート

```python
from ultralytics import YOLO

# 学習済みモデルの読み込み
model = YOLO('yolo26n.pt')

# ONNX形式へのエクスポート
model.export(format='onnx')
```

## 画像サイズを指定したエクスポート

```python
# 推論時に640×640で使う場合
model.export(format='onnx', imgsz=640)

# 推論時に1280×1280で使う場合
model.export(format='onnx', imgsz=1280)
```

## 動的サイズのエクスポート

```python
model.export(format='onnx', imgsz=640, dynamic=True)
```

## ONNXモデルでの推論

```python
from ultralytics import YOLO

# ONNXモデルの読み込み
model = YOLO('yolo26n.onnx')

# 推論の実行
results = model('image.jpg')

# 結果の表示
results[0].show()
```

## OpenVINOへのエクスポート

```python
from ultralytics import YOLO

# 学習済みモデルの読み込み
model = YOLO('yolo26n.pt')

# OpenVINO形式へのエクスポート
model.export(format='openvino')
```

## OpenVINOエクスポートのオプション

```python
# 画像サイズを指定
model.export(format='openvino', imgsz=640)

# 動的サイズを許可
model.export(format='openvino', imgsz=640, dynamic=True)
```

## OpenVINOモデルでの推論

```python
from ultralytics import YOLO

# OpenVINOモデルの読み込み
model = YOLO('yolo26n_openvino_model')

# 推論の実行
results = model('image.jpg')
results[0].show()
```

## PyTorchとOpenVINOの速度比較

```python
from ultralytics.utils.benchmarks import benchmark

# PyTorchモデルでベンチマーク
benchmark(model="yolo26x.pt", format="-", data="coco128.yaml", imgsz=640, device="cpu")

# OpenVINOモデルでベンチマーク
benchmark(model="yolo26x.pt", format="openvino", data="coco128.yaml", imgsz=640, device="cpu")
```
