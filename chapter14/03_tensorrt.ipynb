# 第14章 TensorRT

モデルをTensorRTへ変換し、FP32・FP16・INT8で推論速度を比較します。

## 基本的なTensorRTエクスポート

```python
from ultralytics import YOLO

# 学習済みモデルの読み込み
model = YOLO('yolo26n.pt')

# TensorRT形式へのエクスポート
model.export(format='engine')
```

## TensorRTモデルでの推論

```python
from ultralytics import YOLO

# TensorRTモデルの読み込み
model = YOLO('yolo26n.engine')

# 推論の実行
results = model('image.jpg')
results[0].show()
```

## FP32エクスポート（デフォルト）

```python
# FP32でエクスポート（デフォルト）
model.export(format='engine')

# 明示的に指定する場合
model.export(format='engine', quantize=32)
```

## FP16変換

```python
# FP16変換
model.export(format='engine', quantize=16)
```

## INT8量子化

```python
from ultralytics import YOLO

model = YOLO('yolo26n.pt')

# INT8量子化（キャリブレーションデータが必要）
model.export(
    format='engine',
    quantize=8,
    data='coco128.yaml'  # キャリブレーション用データ
)
```

## データ型別の速度比較

```python
from ultralytics.utils.benchmarks import benchmark

# PyTorchでベンチマーク
benchmark(model='yolo26n.pt', format='-', data='coco128.yaml', device=0)

# TensorRT FP32でベンチマーク
benchmark(model='yolo26n.pt', format='engine', quantize=32, data='coco128.yaml', device=0)

# TensorRT FP16でベンチマーク
benchmark(model='yolo26n.pt', format='engine', quantize=16, data='coco128.yaml', device=0)

# TensorRT INT8でベンチマーク
benchmark(model='yolo26n.pt', format='engine', quantize=8, data='coco128.yaml', device=0)
```
