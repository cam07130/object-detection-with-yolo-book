# 第14章 ベンチマーク

モデル形式ごとの速度と精度をPythonとCLIで比較します。

## 基本的なベンチマーク

```python
from ultralytics.utils.benchmarks import benchmark

# ベンチマークの実行（全形式）
benchmark(model='yolo26n.pt')
```

## 特定形式のベンチマーク

```python
# PyTorchモデルのみベンチマーク
benchmark(model='yolo26n.pt', format='-')

# ONNX形式のみベンチマーク
benchmark(model='yolo26n.pt', format='onnx')
```

## ベンチマークのカスタマイズ

```python
benchmark(
    model='yolo26n.pt',
    data='coco128.yaml',  # 精度評価用データセット（デフォルトはcoco8）
    imgsz=640,            # 画像サイズ
    device=0,             # GPU番号（CPUの場合は'cpu'）
)
```

## CLIによるベンチマーク

```python
# 全形式のベンチマーク
!yolo benchmark model=yolo26n.pt imgsz=640

# 特定形式のみ
!yolo benchmark model=yolo26n.pt format=onnx imgsz=640

# データセット指定
!yolo benchmark model=yolo26n.pt data=coco128.yaml imgsz=640
```
