# 第13章 カスタムデータセットの学習

独自データセットを使った学習と、データが少ない場合のデータ拡張設定を扱います。

## カスタムデータでの学習

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")  # モデルのロード

# 学習の実行
model.train(
    data="path/to/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16
)
```

## 強いデータ拡張の設定

```python
model.train(
    data="data.yaml",
    epochs=100,
    degrees=20.0,      # 回転（デフォルト: 0.0）
    translate=0.2,     # 平行移動（デフォルト: 0.1）
    scale=0.9,         # スケール（デフォルト: 0.5）
    flipud=0.5,        # 上下反転の確率（デフォルト: 0.0）
    mosaic=1.0,        # モザイク拡張（デフォルト: 1.0）
    mixup=0.1          # ミックスアップ（デフォルト: 0.0）
)
```
